# Missing Values in Pandas

**Dataset:** Titanic Dataset


```python
import pandas as pd
import numpy as np

df = pd.read_csv('titanic.csv')
print("Dataset shape:", df.shape)
print("Columns:", list(df.columns))
df.head()
```

## 1. Missing-Value Summary


```python
missing_summary = pd.DataFrame({
    'Missing_Count': df.isna().sum(),
    'Missing_Percentage': (df.isna().mean() * 100).round(2)
}).sort_values('Missing_Count', ascending=False)

missing_summary
```

## 2. Detect Missing Values Using isna() and notna()


```python
# Total missing values
print("Total missing values:", df.isna().sum().sum())

# Missing values in Age
print("Missing Age:", df['age'].isna().sum())

# Rows where Age is available
print("\nRows with Age available:")
print(df.loc[df['age'].notna(), ['name','age']].head())
```

## 3. Dropping Missing Values


```python
# Drop rows where Age is missing
drop_age = df.dropna(subset=['age'])
print("Original rows:", len(df))
print("After dropping missing Age:", len(drop_age))

# Drop rows where Embarked is missing
drop_embarked = df.dropna(subset=['embarked'])
print("After dropping missing Embarked:", len(drop_embarked))
```

## 4. Filling Missing Numerical Values


```python
# Fill Age using the median
cleaned = df.copy()
age_median = cleaned['age'].median()
cleaned['age'] = cleaned['age'].fillna(age_median)

print("Age median used:", age_median)
print("Missing Age after filling:", cleaned['age'].isna().sum())
```

## 5. Filling Missing Categorical Values


```python
# Fill Embarked using its mode
embarked_mode = cleaned['embarked'].mode()[0]
cleaned['embarked'] = cleaned['embarked'].fillna(embarked_mode)

# Fill Cabin with an explicit category
cleaned['cabin'] = cleaned['cabin'].fillna('Unknown')

print("Embarked mode used:", embarked_mode)
print("Missing Embarked after filling:", cleaned['embarked'].isna().sum())
print("Missing Cabin after filling:", cleaned['cabin'].isna().sum())
```

## 6. Before vs After Comparison


```python
comparison = pd.DataFrame({
    'Before': df.isna().sum(),
    'After': cleaned.isna().sum()
})
comparison['Filled_or_Remaining'] = comparison['Before'] - comparison['After']
comparison
```

## 7. Save the Cleaned Dataset


```python
cleaned.to_csv('titanic_cleaned.csv', index=False)
missing_summary.to_csv('missing_value_summary.csv')
print("Saved: titanic_cleaned.csv")
print("Saved: missing_value_summary.csv")
```

## 8. Interview Questions

**1. What are missing values?**  
Missing values are data entries that are not available or were not recorded.

**2. What are common ways to handle missing values?**  
Common methods include dropping rows/columns, filling numerical values with mean/median, and filling categorical values with mode or a meaningful category.

**3. When should you avoid simply dropping missing rows?**  
Avoid dropping rows when many values are missing or when removing those records could reduce the dataset or introduce bias.


## Key Takeaway
- `isna()` and `notna()` help identify missing values.
- `dropna()` removes missing records.
- `fillna()` replaces missing values.
- Median is useful for numerical columns with outliers; mode or an explicit category can be used for categorical columns.
